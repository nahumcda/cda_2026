#### 3.4.3 `for`

Se puede avanzar a la siguiente iteración de un bucle utilizando la *keyword* `continue`.

````python
sequence = [1, 2, None, 4, 5]

total = 0

for value in sequence:
  if value is None:
    continue
  total += value
````

````python
total
````

Un bucle puede abortarse utilizando la *keyword* `break`.

````python
sequence = [1, 2, 0, 4, 6, 5 , 2, 1]

total_until_5 = 0

for value in sequence:
  if value == 5:
    break
  total_until_5 += value
````

#### 3.4.4 Manejo de errores (`try` / `except`)

Cuando una instrucción puede fallar en tiempo de ejecución (por ejemplo, al convertir a número un texto que no lo es), se puede capturar el error con `try` / `except` en lugar de dejar que el programa se detenga:

````python
valores = ['120', '85.5', 'noesunnumero', '40']

for valor in valores:
  try:
    numero = float(valor)
    print(f'{valor} -> {numero}')
  except ValueError:
    print(f'{valor} -> no se ha podido convertir a número')
````

`except ValueError` captura únicamente ese tipo de error; si ocurriera otro tipo de excepción, el programa sí se detendría. Esto evita que un único dato mal formado interrumpa todo el proceso — algo habitual al importar datos reales (por ejemplo, importes de pólizas con un formato incorrecto).

### 4.2 Funciones `lambda` - funciones anónimas  
Funciones simples que constan de una única sentencia, cuyo resultado es el valor de retorno.  
Definida usando la palabra clave `lambda`, que no tiene otro significado que "estamos declarando una función anónima".

````python
def short_function(x):
  return x**2  
````

````python
equiv_anon = lambda x: x**2
````

````python
short_function(3)
````

````python
equiv_anon(3)
````

## 6 Comprensiones de listas, conjuntos y cadenas

Las comprensiones de listas permiten formar de forma concisa una nueva lista filtrando los elementos de una colección y transformando los elementos que pasan el filtro en una expresión concisa.

`[expr for val in collection if condition] `

Equivale a

```
result = []
for val in collection:
  if condition:
    result.append(expr)
```

Para diccionarios y conjuntos de usa la siguientes expresiones:  

`dict_comp = {key_expr : value_expr for value in collection if condition}`

`set_comp = {expr for value in collection if condition}`


````python
lyrics = ['come', 'as', 'you', 'are',',', 'as', 'you', 'were']
````

````python
[cadena for cadena in lyrics]
````

````python
[len(cadena) for cadena in lyrics]
````

````python
{cadena: len(cadena) for cadena in lyrics}
````

````python
{len(cadena) for cadena in lyrics}
````
